```python
import pandas as pd
import matplotlib.pyplot as plt

# Exemple : Création du DataFrame
data = {
    'dateTime': ['2020-01-01', '2020-02-01', '2021-01-01', '2021-02-01', '2022-01-01'],
    'Total active Power': [100, 150, 200, 250, 300]
}
df = pd.DataFrame(data)

# Convertir la colonne dateTime en format datetime
df['dateTime'] = pd.to_datetime(df['dateTime'])

# Extraire les années pour personnaliser les ticks
df['Year'] = df['dateTime'].dt.year

# Tracer le graphique
plt.figure(figsize=(10, 6))
plt.plot(df['dateTime'], df['Total active Power'], marker='o', label='Total active Power')

# Ajuster les étiquettes de l'axe des x
plt.xlabel('Années')
plt.ylabel('Total active Power')
plt.title('Évolution du Total active Power en fonction des années')

# Personnaliser les ticks de l'axe des x pour afficher uniquement les années
years = df['Year'].unique()  # Années uniques
year_dates = df['dateTime'].dt.to_period('Y').drop_duplicates().dt.start_time  # Dates correspondantes au début de l'année
plt.xticks(year_dates, years, rotation=45)

plt.legend()
plt.grid()
plt.tight_layout()
plt.show()
```
